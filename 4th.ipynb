```python
#Dataset
import numpy as np
import pandas as pd
data = {
    "Student Name": ["Ali", "Sara", "Ahmed", "Fatima", "Usman", "Ayesha", "Bilal",
                     "Hina", "Zain", "Maryam", "Hamza", "Noor", "Omar", "Iqra",
                     "Saad", "Laiba", "Danish", "Amna", "Faisal", "Sara"],
    "Age":          [20, 21, 19, 22, 20, 21, 23, 19, 20, 22, 21, 20, 24, 19, 21, 22, 20, np.nan, 23, 21],
    "Gender":       ["Male", "Female", "Male", "Female", "Male", "Female", "Male",
                     "Female", "Male", "Female", "Male", "Female", "Male", "Female",
                     "Male", "Female", "Male", "Female", "Male", "Female"],
    "Study Hours":  [5, 7, 3, 8, 4, 9, 2, 6, np.nan, 8, 5, 7, 1, 6, 4, 9, 3, 7, 2, 7],
    "Attendance":   [85, 92, 70, 95, 78, 98, 60, 88, 72, 94, 82, np.nan, 55, 90, 75, 97, 68, 91, 62, 92],
    "Marks":        [68, 82, 52, 88, 61, 93, 45, 76, 55, 85, 70, 79, 38, 74, 58, 91, 50, 80, 47, 82],
}
df = pd.DataFrame(data)
# Q1. NumPy array of marks: mean, median, std

marks = np.array(df["Marks"])
print("Q1. NumPy statistics of Marks")
print("Marks array :", marks)
print(f"Mean        : {np.mean(marks):.2f}")
print(f"Median      : {np.median(marks):.2f}")
print(f"Std Dev     : {np.std(marks):.2f}\n")
```

```python
# Q2. Pandas DataFrame
print("Q2. DataFrame (first 5 rows)")
print(df.head(), "\n")
print("Shape:", df.shape, "\n")
```

```python
print("Q3. Missing values per column (before)")
print(df.isnull().sum(), "\n")

for col in ["Age", "Study Hours", "Attendance"]:
    df[col] = df[col].fillna(df[col].median())   # fill numeric NaNs with median

print("Missing values per column (after)")
print(df.isnull().sum(), "\n")
```

```python
print("Q4. Duplicate rows found:", df.duplicated().sum())
print(df[df.duplicated(keep=False)], "\n")
df = df.drop_duplicates().reset_index(drop=True)
print("Shape after removing duplicates:", df.shape, "\n")
```

```python
# Q5. Students with Marks > 70

top = df[df["Marks"] > 70]
print("Q5. Students who scored more than 70")
print(top, "\n")
```

```python
# Q6. Seaborn: Study Hours vs Marks

import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")
plt.figure(figsize=(8, 5))
sns.regplot(data=df, x="Study Hours", y="Marks", color="steelblue")
plt.title("Relationship between Study Hours and Marks")
plt.tight_layout()
plt.savefig("q6_study_hours_vs_marks.png", dpi=150)
plt.close()
```

```python
# Q7. Histogram of Marks

import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
sns.histplot(df["Marks"], bins=8, kde=True, color="seagreen")
plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.savefig("q7_marks_histogram.png", dpi=150)
plt.close()

df.to_csv("students_clean.csv", index=False)
print("Saved: q6_study_hours_vs_marks.png, q7_marks_histogram.png, students_clean.csv")
```
